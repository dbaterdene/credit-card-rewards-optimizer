```python
#
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
# 1. Set project paths

BASE_DIR = Path.cwd().parent

REPORTS_DIR = BASE_DIR / "reports"
FIGURES_DIR = BASE_DIR / "figures"

FIGURES_DIR.mkdir(exist_ok=True)
```

```python
# 2. Load calculation results

summary_file = REPORTS_DIR / "card_rewards_summary.csv"
category_file = REPORTS_DIR / "card_rewards_by_category.csv"

summary = pd.read_csv(summary_file)
category_rewards = pd.read_csv(category_file)

print("Summary data loaded successfully.")
print("Summary columns:", summary.columns.tolist())
print("Category columns:", category_rewards.columns.tolist())
```

```python
# 3. Chart 1: Annual Net Value Comparison

summary_sorted = summary.sort_values(
    "ongoing_net_value",
    ascending=True
)

plt.figure(figsize=(10, 6))

plt.barh(
    summary_sorted["card_name"],
    summary_sorted["ongoing_net_value"]
)

plt.title(
    "Annual Net Value by Credit Card",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Annual Net Value ($)")
plt.ylabel("Credit Card")

plt.grid(axis="x", linestyle="--", alpha=0.5)

plt.tight_layout()

plt.savefig(
    FIGURES_DIR / "annual_net_value_comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.close()

print("Chart 1 created.")
```

```python
# 4. Chart 2: Rewards vs. Annual Fee

# Auto-detect the matching column names in `summary`
cols = {c.lower().strip(): c for c in summary.columns}

rewards_col = next((c for k, c in cols.items() if "reward" in k), None)
fee_col = next((c for k, c in cols.items() if "fee" in k), None)
card_col = next((c for k, c in cols.items() if "card" in k or "name" in k), None)

if not all([rewards_col, fee_col, card_col]):
    raise ValueError(f"Could not find expected columns. Available columns are: {list(summary.columns)}")

summary_sorted = summary.sort_values(
    rewards_col, 
    ascending=False
)

plt.figure(figsize=(10, 6))

x = range(len(summary_sorted))
width = 0.35

plt.bar(
    [i - width / 2 for i in x],
    summary_sorted[rewards_col],
    width,
    label="Annual Rewards"
)

plt.bar(
    [i + width / 2 for i in x],
    summary_sorted[fee_col],
    width,
    label="Annual Fee"
)

plt.xticks(
    list(x),
    summary_sorted[card_col],
    rotation=20,
    ha="right"
)

plt.title(
    "Annual Rewards vs. Annual Fees",
    fontsize=16,
    fontweight="bold"
)

plt.ylabel("Amount ($)")
plt.legend()
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()

plt.savefig(
    FIGURES_DIR / "rewards_vs_annual_fees.png",
    dpi=300,
    bbox_inches="tight"
)

plt.close()

print("Chart 2 created.")
```

```python
# 5. Chart 3: Rewards by Spending Category

category_pivot = category_rewards.pivot(
    index="category",
    columns="card_name",
    values="annual_rewards"
)

category_pivot.plot(
    kind="bar",
    figsize=(12, 7)
)

plt.title(
    "Annual Rewards by Spending Category",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Spending Category")
plt.ylabel("Annual Rewards ($)")

plt.xticks(rotation=35, ha="right")

plt.legend(
    title="Credit Card",
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)

plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()

plt.savefig(
    FIGURES_DIR / "category_rewards_comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.close()

print("Chart 3 created.")
```
